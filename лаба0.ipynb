```python
numbers = input('Введи любые целые числа через пробел, я быстро найду их сумму и произведение: ').split()
numbers=list(map(int, numbers))
print("Список введенных чисел=",numbers)
print("Сумма введенных чисел=",sum(numbers))
print("Произведение введенных чисел=",math.prod(numbers))
```
