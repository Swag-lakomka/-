```python
import csv

coefficients = {
    'Песок' : 1.00,
    'Супесь' : 0.67,
    'Глина' : 0.25
}
# Исходный файл с данными
input_csv= 'data.csv'

# Выходной файл с результатами
output_csv= 'result.csv'

# Переменная для хранения результатов
results = []

# Чтение данных из CSV-файла

with open(input_csv, encoding='utf-8') as infile:
    reader = csv.reader(infile)
    next(reader)  # Пропускаем первую строку с заголовком
    for row in reader:
        # Чтение длины, глубины и нижней ширины
        length = float(row)  # Длина траншеи
        depth = float(row)  # Глубина траншеи [1]
        width_bottom = float(row)  # Нижняя ширина траншеи [2]
        # Для каждого типа грунта делаем расчёт
        for ground_type, coeff in coefficients.items():
            # Расчёт верхней ширины траншеи
            width_top = width_bottom + 2 * (depth / coeff)
            # Расчёт объёма земляных работ
            volume = length * depth * ((width_bottom + width_top) / 2)
            # Добавляем результат в список
            result_row = [
                    '%.2f' % length,  # Длина
                    '%.2f' % depth,  # Глубина
                    '% .2f' % width_bottom, # Нижняя ширина
                ground_type,  # Название грунта
                '%.2f' % coeff, # Коэффициент откоса
                '%.2f' % width_top, # Верхняя ширина
                '%.2f' % volume # Объём земляных работ
            ]
            results.append(result_row)
# Запись результатов в CSV-файл
with open(output_csv, 'w', encoding='utf-8', newline='') as outfile:
    writer = csv.writer(outfile)
    # Записываем заголовок
    writer.writerow([
        "Длина траншеи (метры)",
        "Глубина траншеи (метры)",
        "Нижняя ширина траншеи (метры)",
        "Тип грунта",
        "Коэффициент откоса",
        "Верхняя ширина траншеи (метры)",
        "Объём земляных работ (м*)" ])
    # Записываем сами результаты
    writer.writerows(results)

print("Результат успешно сохранён в файл:", output_csv)
```
