```python
import numpy as np
import matplotlib.pyplot as plt

# Задаем границы отрезка и шаг
start = -2  # начало отрезка
finish = 2  # конец отрезка
step = 0.1  # шаг табуляции

# Генерируем массив значений x от -5 до 5
x = np.arange(start, finish + step, step)

# Вычисляем значения функции y = x^3 + 7
y = 2 * x ** 2 - 3 * abs(x) + 4

# Создаем график
plt.figure(figsize=(10, 10))
plt.plot(x, y, label='$y = 2*x**2-3*abs(x)+4$', marker='o', color='red')

# Настраиваем внешний вид графика
plt.title('График функции $y = 2*x**2-3*abs(x)+4$')
plt.xlabel('$x$')
plt.ylabel('$y$')

# Пересечение осей в точке (0,0)
plt.axhline(0, color="black", lw=0.1)
plt.axvline(0, color="black", lw=0.1)

plt.grid(True)
plt.legend()

# Показываем график
plt.show()
```
